```python
#Step 8 – Data Preprocessing
```

```python
df = df.sort_values(by='date')

data = df[['temperature_2m_mean',
           'relative_humidity_2m_mean',
           'wind_speed_10m_max']]

print(data.head())
print(data.shape)
```

```python
#Step 9 –  select only Thiruvananthapuram
```

```python
df = df[df['city'] == 'Thiruvananthapuram']

df = df.sort_values(by='date').reset_index(drop=True)

data = df[['temperature_2m_mean',
           'relative_humidity_2m_mean',
           'wind_speed_10m_max']]

print(data.head())
print(data.shape)
```

```python
#Step 10 – MinMaxScaler (Normalization)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(data)

print(scaled_data[:5])
```

```python
#Step 11 – Create Input Sequences
```

```python
sequence_length = 7

X = []
y = []

for i in range(len(scaled_data) - sequence_length):

    X.append(scaled_data[i:i+sequence_length])

    y.append(scaled_data[i+sequence_length, 0])

X = np.array(X)
y = np.array(y)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#Step 12 – Train, Validation, Test Split
```

```python
total = len(X)

train_size = int(total * 0.70)
val_size = int(total * 0.15)

X_train = X[:train_size]
y_train = y[:train_size]

X_val = X[train_size:train_size + val_size]
y_val = y[train_size:train_size + val_size]

X_test = X[train_size + val_size:]
y_test = y[train_size + val_size:]

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

```python
#Step 13 – SimpleRNN Model
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, Dense, Dropout

model = Sequential()

model.add(Input(shape=(7, 3)))

model.add(SimpleRNN(64))

model.add(Dropout(0.2))

model.add(Dense(1))

model.summary()
```

```python
#Step 14 – Model Compile
```

```python
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

print("Model compiled successfully!")
```

```python
#Step 15 – Model Training
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=32,
    validation_data=(X_val, y_val),
    verbose=1
)
```

```python
#Step — Loss Graph
```

```python
plt.figure(figsize=(10, 5))

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.title('Training vs Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.show()
```

```python
#Part C — Model Evaluation
```

```python
y_pred_scaled = model.predict(X_test)

print("Prediction completed!")
print("Prediction shape:", y_pred_scaled.shape)
```

```python
#next Step — Inverse Scaling
```

```python
y_test_actual = scaler.inverse_transform(
    np.column_stack((y_test, np.zeros((len(y_test), 2))))
)[:, 0]

y_pred_actual = scaler.inverse_transform(
    np.column_stack((y_pred_scaled.flatten(),
                     np.zeros((len(y_pred_scaled), 2))))
)[:, 0]

print("Actual and Predicted temperatures converted to Celsius!")
print("Actual:", y_test_actual[:5])
print("Predicted:", y_pred_actual[:5])
```

```python
#Step — Model Evaluation Metrics
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

rmse = np.sqrt(mean_squared_error(y_test_actual, y_pred_actual))
mae = mean_absolute_error(y_test_actual, y_pred_actual)
r2 = r2_score(y_test_actual, y_pred_actual)

print("RMSE:", rmse)
print("MAE:", mae)
print("R2 Score:", r2)
```

```python
#next Step — Actual vs Predicted Graph
```

```python
plt.figure(figsize=(12, 5))

plt.plot(y_test_actual, label='Actual Temperature')
plt.plot(y_pred_actual, label='Predicted Temperature')

plt.title('Actual vs Predicted Temperature')
plt.xlabel('Test Samples')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.show()
```

```python
#next Step — 7 days Temperature Prediction
```

```python
last_sequence = scaled_data[-7:].copy()

future_predictions_scaled = []

for i in range(7):

    pred = model.predict(
        last_sequence.reshape(1, 7, 3),
        verbose=0
    )[0, 0]

    future_predictions_scaled.append(pred)

    next_row = np.array([
        pred,
        last_sequence[-1, 1],
        last_sequence[-1, 2]
    ])

    last_sequence = np.vstack([
        last_sequence[1:],
        next_row
    ])

future_predictions = scaler.inverse_transform(
    np.column_stack((
        future_predictions_scaled,
        np.zeros((7, 2))
    ))
)[:, 0]

print("Next 7 Days Temperature Forecast:")
print(future_predictions)
```
